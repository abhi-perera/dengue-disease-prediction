## Weekly Coverage

The dengue and weather datasets contain weekly observations. This section checks the number of unique weeks available in each year and identifies any differences in weekly coverage between the two datasets.

```python
dengue_weeks = dengue.groupby("year")["week"].nunique()
print(dengue_weeks)
```

```python
weather_weeks = weather.groupby("year")["week"].nunique()
print(weather_weeks)
```

```python
week_comparison = pd.DataFrame({
    "dengue_weeks": dengue_weeks,
    "weather_weeks": weather_weeks
})

week_comparison
```

```python
dengue_2023_weeks = set(dengue[dengue["year"] == 2023]["week"])

weather_2023_weeks = set(weather[weather["year"] == 2023]["week"])

print("Weeks in weather but not dengue:")
print(weather_2023_weeks - dengue_2023_weeks)

print("Weeks in dengue but not weather:")
print(dengue_2023_weeks - weather_2023_weeks)
```

```python
print("Dengue rows for 2023 week 52:")
print(len(dengue[(dengue["year"] == 2023) & (dengue["week"] == 52)]))

print("Weather rows for 2023 week 52:")
print(len(weather[(weather["year"] == 2023) & (weather["week"] == 52)]))
```

### Observation: Weekly Coverage

The dengue and weather datasets have matching weekly coverage for most years. Years 2016 and 2021 contain 53 weeks in both datasets.

A temporal coverage difference was identified in 2023. Week 52 is present in the weather dataset but absent from the dengue dataset. This accounts for the 26 row difference between the weather and dengue datasets, corresponding to the 26 dengue reporting areas.

This issue is recorded for later preprocessing. No values are added or modified in the raw datasets in this notebook.

## Geographic Coverage

This section examines the geographic units represented in the dengue, weather and population datasets.

Geographic compatibility is important because district names must be standardized before the datasets can be integrated.

```python
dengue_areas = set(dengue["district"].unique())
weather_areas = set(weather["district"].unique())
population_areas = set(population["district"].unique())
```

```python
print("In dengue but NOT in population:")
print(dengue_areas - population_areas)

print("\nIn population but NOT in dengue:")
print(population_areas - dengue_areas)
```

```python
print("In weather but NOT in population:")
print(weather_areas - population_areas)

print("\nIn population but NOT in weather:")
print(population_areas - weather_areas)
```

## Key Findings from the Data Inventory

1. The dengue dataset contains weekly dengue case observations covering 2010–2024.

2. The weather dataset also covers 2010–2024 and contains temperature, humidity, rainfall and wind speed variables.

3. No standard null (`NaN`) values were detected in the three raw datasets. However, temporal completeness must be considered separately from cell level missing values.

4. Dengue Week 52 of 2023 is absent, while the corresponding week exists in the weather dataset.

5. The population dataset covers 2014–2024 rather than the complete 2010–2024 study period.

6. The population dataset contains 200 exact duplicate rows. After excluding exact duplicates for inspection, 275 unique records remain.

7. Geographic naming inconsistencies exist among the datasets, including variations of Hambantota and Nuwara Eliya.

8. The dengue/weather data contain an additional Kalmunai/Kalmune reporting area that requires investigation before integration with the 25 district population dataset.
